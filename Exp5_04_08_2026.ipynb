```python
import pandas as pd
wavelength=[120,119,170,127,121,146,137,190,161,145,141,132]
bins=[118,125,135,160,200]
```

```python
category=pd.cut(wavelength,bins)
category
```

```python
pd.value_counts(category)
```

```python
names=["short wavelength","average wavelength","long wavelength","good wavelength"]
pd.cut(wavelength,bins,labels=names)
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
```

```python
data={'Age': [17,24,21,19,20],'Salary':[3500,4300,6000,13000,6600]}
df=pd.DataFrame(data)
print("original data:\n",df,"\n")
```

```python
min_max_scaler=MinMaxScaler()
df_min_max=pd.DataFrame(
 min_max_scaler.fit_transform(df),
 columns=df.columns
)
print("Min-Max Scaled Data:\n", df_min_max, "\n")
```

```python
standard_scaler=StandardScaler()
df_standard=pd.DataFrame(standard_scaler.fit_transform(df),columns=df.columns)
print("Standardized Data:\n", df_standard, "\n")
```
