```python
import pandas as pd
import numpy as np
```

```python
dffb1=pd.DataFrame({'players': ['messi','lamine','neymar','ronaldo','mbappe','haaland'], 'Goals': [22,7,6,15,19,10]})
dffb2=pd.DataFrame({'players': ['martinez','ferran','cubarsi','williams','bellingham','neymar'],'Goals':[0,8,9,4,3,6]})
```

```python
dfFB1=pd.concat([dffb1,dffb2],ignore_index=True)
dfFB1
```

```python
df=dffb1.merge(dffb2,how='inner')
df
```

```python
df=dffb1.merge(dffb2,how='left')
df
```

```python
df=dffb1.merge(dffb2,how='right')
df
```

```python
data=np.arange(15).reshape((3,5))
indexers=['rainfall','humidity','wind']
dframe1=pd.DataFrame(data,index=indexers,columns=['Bergen','Oslo','Trondheim','Stavanger','Kritiansand'])
dframe1
```

```python
stacked=dframe1.stack()
stacked
stacked.unstack()
```
