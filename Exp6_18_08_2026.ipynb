```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
import pandas as pd
df=pd.read_csv("WineQT.csv")
print(df.head(10))
print(df.tail(10))
```

```python
df.dtypes
```

```python
print(df.describe())
```

```python
pd.isna(df).any()
```

```python
mean=df["density"].mean()
median=df["density"].median()
mode=df["density"].mode()
print(mean,median,mode)
```

```python
df['free sulfur dioxide'].value_counts().nlargest(30).plot(kind='bar', figsize=(10,5))
plt.title("Free Sulphur Dioxide")
plt.xlabel("Free Sulphur Dioxide")
plt.ylabel("Frequency")
```

```python
sns.FacetGrid(df).map(sns.distplot, "free sulfur dioxide").add_legend()
```

```python
sns.FacetGrid(df).map(sns.distplot,"alcohol").add_legend()
```

```python
sns.boxplot(x="free sulfur dioxide",data=df)
plt.show()
```

```python
plt.scatter(df["free sulfur dioxide"], df["pH"])
plt.title("Scatter Plot for pH vs free sulfur dioxide")
plt.xlabel("pH")
plt.ylabel("free sulfur dioxide")
```

```python
sns.boxplot(x="quality",y="free sulfur dioxide",data=df)
plt.show()
```

```python
sns.pairplot(df,vars = ['fixed acidity', 'volatile acidity','citric acid'], kind="reg")
plt.show()
```

```python
sns.set(style="ticks", color_codes=True)
sns.pairplot(df,height=2,vars = ['fixed acidity', 'volatile acidity','citric acid'], hue="quality")
plt.show()
```

```python
from scipy import stats
corr=stats.pearsonr(df['free sulfur dioxide'],df['quality'])
print("value:\t",corr[1])
print("cor:\t\t",corr[0])
```

```python
correlation = df.corr(method='pearson', numeric_only=True)
correlation
```

```python
sns.heatmap(correlation,xticklabels=correlation.columns,
 yticklabels=correlation.columns)
```
