```python
import pandas as pd
import numpy as np
```

```python
dataframe1=pd.DataFrame({'StudentID':[1,3,5,7,9], 'Score':[89,39,50,97,22]})
dataframe2=pd.DataFrame({'StudentID':[2,4,6,8,10], 'Score':[98,93,44,77,69]})
```

```python
vertical_concat=pd.concat([dataframe1,dataframe2],ignore_index=True)
print("Vertical Concatenation:")
print(vertical_concat)
horizontal_concat=pd.concat([dataframe1,dataframe2],axis=1)
print("\nHorizontal Concatenation:")
print(horizontal_concat)
```

```python
dfSE=pd.DataFrame({'StudentID':[2,4,6,8,10,12],'ScoreSE':[98,93,44,77,69,56]})
dfML=pd.DataFrame({'StudentID':[1,2,3,4,5,6,8,10],'ScoreML':[39,93,49,44,55,78,97,87]})
```

```python
inner_join=pd.merge(dfSE,dfML,on='StudentID',how='inner')
print("\ninner join:")
print(inner_join)
left_join=pd.merge(dfSE,dfML,on='StudentID',how='left')
print("\nleft join:")
print(left_join)
right_join=pd.merge(dfSE,dfML,on='StudentID',how='right')
print("\nright join:")
print(right_join)
outer_join=pd.merge(dfSE,dfML,on='StudentID',how='outer')
print("\nouter join:")
print(outer_join)
```

```python
sales=pd.DataFrame({'Quantity':[100,250,300,150],'UnitPrice':[50,60,45,80]})
sales['TotalPrice']=sales['Quantity']*sales['UnitPrice']
print("\nSales Data:")
print(sales)
print("\nDescriptive Statistics:")
print(sales.describe())
company_df=pd.DataFrame({'Company':['ABC','XYZ','ABC','PQR','XYZ','ABC']})
print("\nCompany Counts:")
print(company_df['Company'].value_counts())
```

```python
data=np.arange(15).reshape((3,5))
dframe=pd.DataFrame(data,index=['Rainfall','Humidity','Wind'],columns=['Bergen','Oslo','Trondheim','Stavanger','Kristiansand'])
print("\nOriginal DataFrame:")
print(dframe)
stacked=dframe.stack()
print(dframe)
stacked=dframe.stack()
print("\nStacked Data:")
print(stacked)
unstacked=stacked.unstack()
print("\nUnstacked Data:")
print(unstacked)
```

```python
series1=pd.Series([0,111,222,333],index=['zero','ones','twos','threes'])
series2=pd.Series([444,555,666],index=['fours','fives','sixs'])
frame=pd.concat([series1,series2],keys=['Number1','Number2'])
print("\nHierarchial Series:")
print(frame)
print("\nUnstacked Series:")
print(frame.unstack())
```
